# Matrix Calculus: The Linear Operator View
we view derivatives as linear operators that map a small change in input ($dX$) to a change in output ($dY$).
For a simple layer $Y = XW$, the differential is:
$$dY = (dX)W$$
This means the sensitivity of the output depends entirely on the weights $W$.

```julia
using LinearAlgebra

# 1. Setup our "Layer"
X = [1.0 2.0; 3.0 4.0]  # Input
W = [0.5 0.1; 0.2 0.8]  # Weights (Parameters)
Y = X * W               # Initial Output

# 2. Define a "Small Perturbation" (dX)
# Imagine we change the input slightly
dX = [0.01 0.0; 0.0 0.01] 

# 3. Calculate the actual change in output
Y_new = (X + dX) * W
actual_dY = Y_new - Y

# 4. Calculate dY using the Differential Rule: dY = (dX)W
theoretical_dY = dX * W

println("Actual change (Y_new - Y):")
display(actual_dY)

println("\nTheoretical change (dX * W):")
display(theoretical_dY)

println("\nAre they equal? ", isapprox(actual_dY, theoretical_dY))
```

```julia
# Create a random Orthogonal Matrix Q
# In Julia, 'qr' decomposition is a quick way to get one
A = randn(2, 2)
Q, R = qr(A)
Q = Matrix(Q) # This is our Orthogonal W

# Check: Is Q'Q = I?
println("Is Q orthogonal? ", isapprox(Q' * Q, I))

# Observe: Does Q preserve the magnitude of dX?
println("Norm of dX: ", norm(dX))
println("Norm of (dX * Q): ", norm(dX * Q))
```
