We will test basic matrix calculus operations by computing the Jacobian of the softmax function for a single row of attention scores.
The mathematical derivation is:
$$J = \text{diag}(p) - p p^T$$

```julia
using LinearAlgebra

function manual_softmax_jacobian(s)
    # Compute the softmax probabilities
    exp_s = exp.(s)
    p = exp_s ./ sum(exp_s)
    
    # Construct the Jacobian matrix: diag(p) - p * p^T
    # In Julia, p * p' automatically handles the outer product for vectors
    J = Diagonal(p) - p * p'
    
    return J
end
```

### Execution and Verification
Next, we generate a random input vector $s$ of size 5 to simulate a small set of pre-softmax logits, and compute the resulting $5 \times 5$ Jacobian matrix.

```julia
import Random
Random.seed!(42)

# Initialize a random float64 vector 
s = randn(Float64, 5)
println("Input vector s:")
display(s)

# Compute the theoretical Jacobian
J_manual = manual_softmax_jacobian(s)

println("\nCalculated Jacobian Matrix:")
display(J_manual)
```
